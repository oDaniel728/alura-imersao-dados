```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    "nome": ["Ana", "Bruno", "Carlos", "Daniel"],
    "salario": [4000, 5000, np.nan, 4500]
})
```

```python
df["salario"] = df["salario"]\
    .fillna(df["salario"].median())\
    .round(2)

## Pegar o nome da pessoa com menor salário
# df[df["salario"] == df["salario"].min()]
# df.nsmallest(1, "salario")
# df.query("salario == salario.min()")
```
