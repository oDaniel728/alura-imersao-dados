```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame({
    "dia": ["Domingo", "Segunda", "Terça", "Quarta", "Quinta", "Sexta", "Sábado"],
    "temperatura": [30.0, 27.0, 31.0, 36.0, 38.5, 35.0, 32.0],
    "prob_chuva":  [80  , 70  , 65  , 32  , 12  , 21  , 33  ],

})
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    df["dia"],
    df["temperatura"],
    color="red",
    marker="o",
    label="Temperatura",
    markerfacecolor="white",
    markeredgecolor="red",
    markersize=8,
    markeredgewidth=2,
)
plt.plot(df["dia"], df["prob_chuva"], color="blue", marker="o", label="Prob. Chuva")

yoffset = (-7, 4)
xoffset = (0, 0)
font_size = 12
for i, (dia, temp) in enumerate(zip(df["dia"], df["temperatura"])):
    plt.text(
        i + xoffset[0],
        temp + yoffset[0],
        f"{temp:.0f}ºC",
        ha="center",
        va="bottom",
        color="black",
        fontsize=font_size,
    )

for i, (dia, prob) in enumerate(zip(df["dia"], df["prob_chuva"])):
    plt.text(
        i + xoffset[1],
        prob + yoffset[1],
        f"{prob}%",
        ha="center",
        va="bottom",
        color="black",
        fontsize=font_size,
    )

plt.xlabel("Dia da semana")
plt.ylabel("Valores")
plt.title("Temperatura e Probabilidade de Chuva")
plt.legend()
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```
