# Aula 1

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/guilhermeonrails/data-jobs/refs/heads/main/salaries.csv")
```

```python
df.head()
```

```python
df.describe()
```

```python
rows, columns = df.shape
print(f"Linhas: {rows};\nColunas: {columns}.")
```

```python
df.columns
```

```python
traducao = {
    "work_year": "ano_de_trabalho",
    "experience_level": "nivel_de_experiencia",
    "employment_type": "tipo_de_emprego",
    "job_title": "cargo",
    "salary": "salario",
    "salary_currency": "moeda_do_salario",
    "salary_in_usd": "salario_em_usd",
    "employee_residence": "residencia_do_funcionario",
    "remote_ratio": "percentual_remoto",
    "company_location": "localizacao_da_empresa",
    "company_size": "tamanho_da_empresa",
}

df.rename(columns=traducao, inplace=True)
```

```python
traducao = {
    "SE": "senior",
    "MI": "pleno",
    "EN": "junior",
    "EX": "executivo"
}
df["nivel_de_experiencia"] = df["nivel_de_experiencia"].astype(str).replace(traducao)
```

```python
df["tipo_de_emprego"] = df["tipo_de_emprego"].astype(str).replace({
    "FT": "integral",
    "PT": "parcial",
    "FL": "freelancer",
    "CT": "contrato"
})
```

```python
df["tamanho_da_empresa"] = df["tamanho_da_empresa"].astype(str).replace({
    "S": "pequena",
    "M": "média",
    "L": "grande"
})
```

```python
df["percentual_remoto"] = df["percentual_remoto"].replace(
    {100: "remoto", 50: "parcial", 0: "presencial"}
)
```

```python
df.describe(include="all")
```

```python
df.head()
```

```python

salarios = df.groupby("nivel_de_experiencia")["salario"].mean().sort_values(ascending=False).index
```

```python
plt.figure(figsize=(4, 5))
sns.barplot(data=df, x='nivel_de_experiencia', y='salario', order=salarios)
plt.title("Salário médio por nível de experiência")

plt.xlabel("Nível de experiência")
plt.ylabel("Salário")

plt.show()
```

```python
senioridades: list[str] = df["nivel_de_experiencia"].unique().tolist()
plt.figure(figsize=(10, 5))
sns.boxplot(x=df["nivel_de_experiencia"], y='salario', data=df, order=senioridades)
plt.title("Salário médio por nível de experiência")

plt.xlabel("Nível de experiência")
plt.ylabel("Salário")

plt.show()
```
